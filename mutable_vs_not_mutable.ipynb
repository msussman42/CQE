```python
def fact_mutable(nlist):
    nlist_new=nlist.copy()
    if nlist[0]== 1:
        return 1
    else:
        nlist_new[0]=nlist[0]-1
        y=fact_mutable(nlist_new)
        return nlist[0] * y
nlist=[5]
x = fact_mutable(nlist)
print(x)
```

```python
def fact_bad_mutable(nlist):
    if nlist[0] == 1:
        return 1
    else:
        nlist[0]=nlist[0]-1
        y=fact_bad_mutable(nlist)
        return (nlist[0]+1) * y
nlist=[5]
x = fact_bad_mutable(nlist)
print(x)
```

```python
def fact_not_mutable(n):
    if n == 1:
        return 1
    else:
        n=n-1
        y=fact_not_mutable(n)
        return (n+1) * y
n=5
x = fact_not_mutable(n)
print(x)
```

```python
def fact_universal(n):
    if n == 1:
        return 1
    else:
        ncopy=n-1
        y=fact_universal(ncopy)
        return n * y
n=5
x = fact_universal(n)
print(x)
```
