```python
import pylab
import numpy as np
theta = np.linspace(0.0001, np.pi, 10000)
alambda=0.9
a=1.0
numerator=alambda*np.sin(theta)
denominator=1.0-2.0*(alambda**2)*(np.sin(theta*0.5)**2)
#k/h=lambda k=h lambda.        psi k =theta lambda=theta alambda/a
alpha=np.atan(numerator/denominator)*a/(theta*alambda)
fig2 = pylab.figure()
ax2 = fig2.add_subplot(111)
ax2.plot(theta, alpha)
pylab.show()
```
