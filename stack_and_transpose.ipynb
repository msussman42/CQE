```python
import numpy as np
a=np.array([[1,5,10]])
b=np.array([[2,4,11]])
atran=a.transpose()
btran=b.transpose()
print(a)
print(b)
print(atran)
print(btran)
Y=np.vstack((atran,btran))
print(Y)
YT=np.vstack((atran,btran)).T
print(YT)
n=10
for i in range(n):
  print(i)
```
